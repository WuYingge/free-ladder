## 更新A股财务数据 (financial)

七张宽表落盘 `data/financial/`（income_q/balance_q/cashflow_q/forecast/share_capital/dividend/holder_num）。
口径：金额=元、股本=股、比率=小数、金额累计口径；`ann_date`=首次公告日（PIT 关键：公告日之前该期数据不可见）。
披露季（4/8/10 月）建议每天跑；非披露季可只更 `forecast,dividend`。
日更成本：事件表 <1 分钟；三张报表约 5355 只 × 2 报告期 ÷ 8 线程 ≈ 4~5 分钟。
首次回填请用 CLI：`python libs/scripts/update_financials.py --all-history --start-year 2014`。


```python
os.makedirs(DataPath.FINANCIAL_DIR, exist_ok=True)
sync_backup(src=DataPath.FINANCIAL_DIR, dest=os.path.join(DataPath.BAK_PATH, "financial"))
```
